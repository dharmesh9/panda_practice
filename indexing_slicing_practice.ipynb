```python
import pandas as pd
import numpy as np
```

```python

# Write a Pandas program to select rows where the value in the 'A' column is greater than 4.

df = pd.DataFrame({
    'A': [1, 6, 8, 3, 7],
    'B': [5, 2, 9, 4, 1]
})

# Select rows where 'A' is greater than 4
result = df[df['A'] > 4]
print(result)
```

```python

# Write a Pandas program to select only the 'X' and 'Y' columns from the DataFrame.

# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1],
    'Z': [3, 8, 6, 2, 9]
})

# Select 'X' and 'Y' columns
result = df[['X', 'Y']]
print(result)
```

```python

# Write a Pandas program to set a MultiIndex and access specific data using it.

# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1],
    'Z': ['one', 'one', 'two', 'two', 'one']
})

# Set MultiIndex
df = df.set_index(['Z', 'X'])

# Access data using MultiIndex
result = df.loc[('one', 6)]

print(result)
```

```python

# Write a Pandas program to slice DataFrame based on MultiIndex levels.

# Create a DataFrame
df = pd.DataFrame({
    'A': [1, 6, 8, 3, 7],
    'B': [5, 2, 9, 4, 1],
    'C': ['one', 'one', 'two', 'two', 'one']
})

# Set MultiIndex
df = df.set_index(['C', 'A'])

# Slice DataFrame
result = df.loc['one']
print(result)
```

```python
# Write a Pandas program to swap the levels of a MultiIndex DataFrame.


# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1],
    'Z': ['one', 'one', 'two', 'two', 'one']
})

# Set MultiIndex
df = df.set_index(['Z', 'X'])

# Swap the levels of the MultiIndex
df = df.swaplevel()
print(df)
```

```python
# Write a Pandas program to reset the index of a MultiIndex DataFrame.

# Create a DataFrame
df = pd.DataFrame({
    'A': [1, 6, 8, 3, 7],
    'B': [5, 2, 9, 4, 1],
    'C': ['one', 'one', 'two', 'two', 'one']
})

# Set MultiIndex
df = df.set_index(['C', 'A'])

# Reset the index
df = df.reset_index()
print(df)
```

```python
# Write a Pandas program that uses .loc for indexing.


# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1]
})

# Indexing using .loc
result = df.loc[1]
print(result)
```

```python
# Write a Pandas program to select the first three rows using iloc.

# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1]
})

# Select first three rows using iloc
result = df.iloc[:3]
print(result)
```

```python
# Write a Pandas program to use .loc to select rows based on a condition.


# Create a DataFrame
df = pd.DataFrame({
    'P': [1, 6, 8, 3, 7],
    'Q': [5, 2, 9, 4, 1]
})

# Conditional selection using .loc
result = df.loc[df['P'] > 3]
print(result)
```

```python
# Write a Pandas program that uses .loc to set values in the DataFrame.

# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1]
})

# Set values using .loc
df.loc[df['X'] > 5, 'Y'] = 0
print(df)
```

```python
# Write a Pandas program that uses .loc to slice DataFrame based on row and column labels.


# Create a DataFrame
df = pd.DataFrame({
    'X': [1, 6, 8, 3, 7],
    'Y': [5, 2, 9, 4, 1],
    'Z': [7, 8, 9, 1, 2]
})

# Slice DataFrame using .loc
result = df.loc[1:3, ['X', 'Z']]
print(result)
```

```python
# Write a Pandas program that uses .loc to slice a MultiIndex DataFrame.

# Create a DataFrame
df = pd.DataFrame({
    'A': [1, 6, 8, 3, 7],
    'B': [5, 2, 9, 4, 1],
    'C': ['one', 'one', 'two', 'two', 'one']
})

# Set MultiIndex
df = df.set_index(['C', 'A'])

# Sort the MultiIndex
df = df.sort_index()

# Slice MultiIndex DataFrame using .loc
result = df.loc['one':'two']
print(result)
```
