```python
# Write a Pandas program to add, subtract, multiple and divide two Pandas Series.
# Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]


ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 9])

print("Add two Series:")
ds = ds1 + ds2
print(ds)

print("Subtract two Series:")
ds = ds1 - ds2
print(ds)

print("Multiply two Series:")
ds = ds1 * ds2
print(ds)

print("Divide Series1 by Series2:")
ds = ds1 / ds2
print(ds)
```

```python
# Write a Pandas program to compare the elements of the two Pandas Series.
# Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 10]

ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 10])

print("Series1:")
print(ds1)

print("Series2:")
print(ds2)

print("Compare the elements of the said Series:")
print("Equals:")
print(ds1 == ds2)

print("Greater than:")
print(ds1 > ds2)

print("Less than:")
print(ds1 < ds2)
```

```python
# Flatten Series of Lists

# Write a Pandas program to convert Series of lists to one Series.

# Sample Output: 
# Original Series of list
# 0    [Red, Green, White]
# 1           [Red, Black]
# 2               [Yellow]
# dtype: object
# One Series
# 0       Red
# 1     Green
# 2     White
# 3       Red
# 4     Black
# 5    Yellow
# dtype: object

s = pd.Series([
    ['Red', 'Green', 'White'],
    ['Red', 'Black'],
    ['Yellow']])
print("Original Series of list")
print(s)
s = s.explode().reset_index(drop=True)
print("One Series")
print(s)
```

```python
# Write a Pandas program to change the order of index of a given series.

# Sample Output:
# Original Data Series:
# A    1
# B    2
# C    3
# D    4
# E    5
# dtype: int64
# Data Series after changing the order of index:
# B    2
# A    1
# C    3
# D    4
# E    5
# dtype: int64


s1 = pd.Series(data = [1,2,3,4,5], index = ['A', 'B', 'C','D','E'])

print("Original Data Series:")
print(s1)

s2 = s1.reindex(index = ['B','A','C','D','E'])
print("\nData Series after changing the order of index:")
print(s2)
```

```python
# Write a Pandas program to get the items of a given series not present in another given series.
# Original Series:
# sr1:
# 0    1
# 1    2
# 2    3
# 3    4
# 4    5

# sr2:
# 0     2
# 1     4
# 2     6
# 3     8
# 4    10


# Items of sr1 not present in sr2:
# 0    1
# 2    3
# 4    5


sr1 = pd.Series([1, 2, 3, 4, 5])
sr2 = pd.Series([2, 4, 6, 8, 10])
print("Original Series:")
print("sr1:")
print(sr1)
print("\nsr2:")
print(sr2)
print("\nItems of sr1 not present in sr2:")

sr3 = sr1[~sr1.isin(sr2)]
print(sr3)
```

```python
# Write a Pandas program to get the items which are not common of two given series.
# Original Series:
# sr1:
# 0    1
# 1    2
# 2    3
# 3    4
# 4    5

# sr2:
# 0     2
# 1     4
# 2     6
# 3     8
# 4    10


# Items of a given series not present in another given series:
# 0     1
# 2     3
# 4     5
# 5     6
# 6     8
# 7    10

sr1 = pd.Series([1, 2, 3, 4, 5])
sr2 = pd.Series([2, 4, 6, 8, 10])
print("Original Series:")
print("sr1:")
print(sr1)
print("sr2:")
print(sr2)
print("\nItems of a given series not present in another given series:")
sr11 = pd.Series(np.union1d(sr1, sr2))
print('\n Union of both series')
print(sr11)
sr22 = pd.Series(np.intersect1d(sr1, sr2))
print('\n Intersect of both series')
print(sr22)

result = sr11[~sr11.isin(sr22)]
print('\n')
print(result)
```

```python
# Write a Pandas program to display most frequent value in a given series and replace everything else as ‘Other’ in the series.
# Original Series:
# 0     3
# 1     1
# 2     1
# 3     3
# 4     2
# 5     2
# 6     1
# 7     2
# 8     3
# 9     1
# 10    2
# 11    2
# 12    2
# 13    3
# 14    3
# dtype: int64
# Top 2 Freq: 2    6
# 3    5
# 1    4
# dtype: int64
# 0     Other
# 1     Other
# 2     Other
# 3     Other
# 4         2
# 5         2
# 6     Other
# 7         2
# 8     Other
# 9     Other
# 10        2
# 11        2
# 12        2
# 13    Other
# 14    Other


num_series = pd.Series(np.random.randint(1, 5, 15))

print("Original Series:")
print(num_series)

freq = num_series.value_counts()
print("Top 2 Freq:")
print(freq)

top1 = freq.index[0]

# convert dtype so we can store strings
result = num_series.astype(object)

result[~num_series.isin([top1])] = 'Other'

print(result)
```
